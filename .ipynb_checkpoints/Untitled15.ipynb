```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Sample - Superstore.csv",encoding="latin1")
print("Dataset loaded successfully") 
```

```python
print("=====================================================First 5 Rows====================================================")
print(df.head())
print("=====================================================Last 5 Rows======================================================")
print(df.tail())
print("=====================================================Sample Records======================================================")
print(df.sample())
print("=====================================================Dataset Info======================================================")
print(df.info())
print("=====================================================Dataset Shape======================================================")
print(df.shape)
print("=====================================================Columns Names======================================================")
print(df.columns.tolist())
```

```python
print("=====================================================Missing Values======================================================")
print(df.isnull().sum())
print("=====================================================Duplicate Records======================================================")
print(df.duplicated().sum())
print("=====================================================Data Types======================================================")
print(df.dtypes)
```

```python
df["Order Date"]=pd.to_datetime(df["Order Date"])
print(df["Order Date"].head())
print(df["Order Date"].dtype)
```

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
```

```python
print(df.describe())
```

```python
total_sales=df["Sales"].sum()
total_profit=df["Profit"].sum()
average_discount=df["Discount"].mean()
print("Total Sales:",round(total_sales,2))
print("Total Profit:",round(total_profit,2))
print("Average Discount:",round(average_discount,2))
```

```python
sales=np.array(df["Sales"])
profit=np.array(df["Profit"])
discount=np.array(df["Discount"])
print("Average Sales:",np.mean(sales))
print("Median Sales:",np.median(sales))
print("Minimum Sales:",np.min(sales))
print("Maximum Sales:",np.max(sales))
print("Average Profit:",np.mean(profit))
print("Median Profit:",np.median(profit))
print("Minimum Profit:",np.min(profit))
print("Maximum Profit:",np.max(profit))
print("Average Discount:",np.mean(discount))
```

```python
category=df.groupby("Category")[["Sales","Profit"]].sum()
print(category)
```

```python
print("Most profitable category=",category["Profit"].idxmax())
print("Least profitable category=",category["Profit"].idxmin())
```

```python
yearly=df.groupby("Year")[["Sales", "Profit"]].sum()
print(yearly)
```

```python
monthly=df.groupby("Month")[["Sales", "Profit"]].sum()
print(monthly)
```

```python
correlation=df[["Discount", "Sales", "Profit"]].corr()
print(correlation)
```

```python
region=df.groupby("Region")[["Sales", "Profit"]].sum()
print(region)
print("Most Profitable Region:",region["Profit"].idxmax())
print("Least Profitable Region:",region["Profit"].idxmin())
```

```python
segment=df.groupby("Segment")[["Sales","Profit"]].sum()
print(segment)
print("Most Profitable Segment:",segment["Profit"].idxmax())
print("Least Profitable Segment:",segment["Profit"].idxmin())
```

```python
products = df.groupby("Product Name")[["Sales", "Profit"]].sum()
print(products)
average_sales = products["Sales"].mean()
problem_products = products[(products["Sales"] > average_sales) &(products["Profit"] < 0)]
print(problem_products)
```

```python
category["Profit"].plot(kind="bar")
plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.show()
```

```python
yearly.plot(kind="line", marker="o")
plt.title("Yearly Sales and Profit Trend")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.show()
```

```python
monthly.plot(kind="line", marker="o")
plt.title("Monthly Sales and Profit Trend")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.show()
```

```python
sns.scatterplot(data=df, x="Discount", y="Profit")
plt.title("Discount vs Profit")
plt.show()
```

```python
sns.scatterplot(data=df, x="Discount", y="Sales")
plt.title("Discount vs Sales")
plt.show()
```

```python
sns.heatmap(correlation, annot=True)
plt.title("Correlation Between Discount, Sales and Profit")
plt.show()
```

```python
region["Profit"].plot(kind="bar")
plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.show()
```

```python
segment["Profit"].plot(kind="bar")
plt.title("Profit by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Profit")
plt.show()
```

```python
top_10_loss = problem_products.nsmallest(10, "Profit")
top_10_loss["Profit"].plot(kind="bar")
plt.title("Top 10 High Sales Products with Negative Profit")
plt.xlabel("Product")
plt.ylabel("Profit")
plt.xticks(rotation=90)
plt.show()
```

```python
sns.histplot(df["Profit"], bins=30)
plt.title("Profit Distribution")
plt.xlabel("Profit")
plt.show()
```

```python
df.to_csv("Cleaned_Superstore.csv",index=False)
print("Cleaned Dataset Saved Successfully")
```
